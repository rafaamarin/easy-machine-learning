# Series y DataFrames en Pandas

**Pandas** es la biblioteca fundamental para el análisis de datos en Python. Su nombre proviene de *Panel Data*, un término utilizado en econometría para datos multidimensionales. Antes de manipular o analizar cualquier conjunto de datos, es esencial comprender las dos estructuras básicas que Pandas proporciona: **Series** y **DataFrame**.

Estas estructuras permiten representar, organizar y manipular datos de forma eficiente, aprovechando operaciones vectorizadas que evitan la necesidad de escribir bucles explícitos.

## Series: datos unidimensionales

Una **Series** es una estructura de datos unidimensional que puede almacenar cualquier tipo de dato: enteros, flotantes, cadenas de texto, objetos Python, entre otros. Cada elemento de una Series tiene asociado un **índice** que permite acceder a él de forma directa.

```mermaid
flowchart LR
    subgraph Series
        direction TB
        I0[índice 0] --> V0[valor A]
        I1[índice 1] --> V1[valor B]
        I2[índice 2] --> V2[valor C]
    end
```

### Creación de una Series

La forma más común de crear una Series es a partir de una lista de Python:

````python
import pandas as pd

# Crear una Series a partir de una lista
temperaturas = pd.Series([22.5, 24.1, 19.8, 23.0, 21.5])
print(temperaturas)
````

El resultado muestra tanto el índice como los valores:

```
0    22.5
1    24.1
2    19.8
3    23.0
4    21.5
dtype: float64
```

> Una Series siempre tiene un **índice**. Si no se especifica uno, Pandas genera automáticamente un índice numérico que comienza en 0.

### Series con índice personalizado

En muchos escenarios resulta útil asignar etiquetas significativas al índice:

````python
# Series con índice personalizado
ventas_diarias = pd.Series(
    [1500, 2300, 1800, 2100, 2500],
    index=['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
)
print(ventas_diarias)
````

Salida:

```
Lunes        1500
Martes       2300
Miércoles    1800
Jueves       2100
Viernes      2500
dtype: int64
```

### Series a partir de un diccionario

Los diccionarios de Python se convierten de forma natural en Series, donde las claves se transforman en el índice:

````python
# Crear Series desde un diccionario
poblacion = pd.Series({
    'Madrid': 3300000,
    'Barcelona': 1600000,
    'Valencia': 800000,
    'Sevilla': 690000
})
print(poblacion)
````

### Atributos fundamentales de Series

Las Series exponen varios atributos que permiten inspeccionar su estructura:

````python
# Atributos básicos de una Series
print(ventas_diarias.values)    # Array NumPy con los valores
print(ventas_diarias.index)     # Índice de la Series
print(ventas_diarias.dtype)     # Tipo de dato de los elementos
print(ventas_diarias.shape)     # Forma: (número de elementos,)
print(ventas_diarias.size)      # Cantidad total de elementos
print(ventas_diarias.name)      # Nombre de la Series (None si no tiene)
````

> El atributo **dtype** indica el tipo de dato que almacena la Series. Pandas infiere automáticamente el tipo más adecuado, pero también permite especificarlo de forma explícita.

### Asignar nombre a una Series

Una Series puede tener un nombre, lo cual resulta útil cuando forma parte de un DataFrame:

````python
# Series con nombre
ingresos = pd.Series([45000, 52000, 48000], name='ingresos_mensuales')
print(ingresos.name)  # ingresos_mensuales
````

## DataFrame: datos bidimensionales

Un **DataFrame** es una estructura bidimensional que organiza los datos en filas y columnas, similar a una tabla de base de datos o una hoja de cálculo. Cada columna de un DataFrame es, internamente, una Series.

```mermaid
flowchart TB
    subgraph DataFrame
        direction TB
        subgraph Columnas
            C1[nombre]
            C2[edad]
            C3[ciudad]
        end
        subgraph Filas
            R0[fila 0]
            R1[fila 1]
            R2[fila 2]
        end
    end
    C1 --> R0
    C2 --> R0
    C3 --> R0
```

### Creación de un DataFrame

La forma más habitual de crear un DataFrame es mediante un diccionario donde las claves representan los nombres de las columnas:

````python
# Crear DataFrame desde un diccionario
datos = {
    'nombre': ['Ana', 'Carlos', 'Elena', 'David'],
    'edad': [28, 35, 42, 31],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla'],
    'salario': [45000, 52000, 61000, 48000]
}

df = pd.DataFrame(datos)
print(df)
````

Resultado:

```
   nombre  edad     ciudad  salario
0     Ana    28     Madrid    45000
1  Carlos    35  Barcelona    52000
2   Elena    42   Valencia    61000
3   David    31    Sevilla    48000
```

> Un DataFrame tiene dos índices: uno para las **filas** (índice de fila) y otro para las **columnas** (nombres de columna). Ambos permiten acceder a los datos de forma estructurada.

### DataFrame con índice personalizado

Al igual que las Series, los DataFrames pueden usar índices personalizados:

````python
# DataFrame con índice personalizado
empleados = pd.DataFrame(
    {
        'departamento': ['Ventas', 'IT', 'RRHH', 'Marketing'],
        'empleados': [15, 23, 8, 12],
        'presupuesto': [150000, 280000, 95000, 120000]
    },
    index=['D001', 'D002', 'D003', 'D004']
)
print(empleados)
````

### DataFrame desde lista de diccionarios

Otra forma común de construir DataFrames es a partir de una lista de diccionarios, donde cada diccionario representa una fila:

````python
# Lista de registros (cada dict es una fila)
registros = [
    {'producto': 'Laptop', 'precio': 899, 'stock': 45},
    {'producto': 'Monitor', 'precio': 299, 'stock': 120},
    {'producto': 'Teclado', 'precio': 79, 'stock': 200}
]

inventario = pd.DataFrame(registros)
print(inventario)
````

## Atributos esenciales del DataFrame

Los DataFrames proporcionan múltiples atributos para examinar su estructura y contenido. Conocerlos resulta fundamental para trabajar de forma efectiva con los datos.

### Dimensiones y forma

````python
# Crear DataFrame de ejemplo
df = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D', 'E'],
    'ventas': [100, 150, 200, 175, 225],
    'precio': [10.5, 15.0, 12.0, 18.5, 20.0],
    'categoria': ['X', 'Y', 'X', 'Z', 'Y']
})

# Forma del DataFrame (filas, columnas)
print(df.shape)  # (5, 4)

# Número de dimensiones
print(df.ndim)  # 2

# Total de elementos
print(df.size)  # 20
````

> El atributo **shape** devuelve una tupla con el número de filas y columnas. Es la forma más rápida de conocer las dimensiones de un conjunto de datos.

### Columnas e índice

````python
# Nombres de las columnas
print(df.columns)
# Index(['producto', 'ventas', 'precio', 'categoria'], dtype='object')

# Índice de las filas
print(df.index)
# RangeIndex(start=0, stop=5, step=1)
````

### Tipos de datos

El atributo **dtypes** muestra el tipo de dato de cada columna:

````python
print(df.dtypes)
````

Salida:

```
producto     object
ventas        int64
precio      float64
categoria    object
dtype: object
```

> El tipo **object** generalmente representa cadenas de texto. Pandas también soporta el tipo **string** más moderno, optimizado para operaciones con texto.

### Valores subyacentes

El atributo **values** devuelve los datos como un array de NumPy:

````python
print(df.values)
````


## Inspección rápida de datos

Cuando se trabaja con conjuntos de datos reales, rara vez se visualiza el DataFrame completo. Pandas ofrece métodos para inspeccionar rápidamente la estructura y el contenido de los datos.

### Visualizar primeras y últimas filas

Los métodos **head** y **tail** permiten ver un subconjunto de filas:

````python
# Crear DataFrame más grande para demostración
import numpy as np

np.random.seed(42)
datos_ventas = pd.DataFrame({
    'fecha': pd.date_range('2025-01-01', periods=100),
    'producto': np.random.choice(['A', 'B', 'C'], 100),
    'cantidad': np.random.randint(1, 50, 100),
    'precio_unitario': np.random.uniform(10, 100, 100).round(2)
})

# Primeras 5 filas (por defecto)
print(datos_ventas.head())

# Primeras 3 filas
print(datos_ventas.head(3))

# Últimas 5 filas (por defecto)
print(datos_ventas.tail())

# Últimas 10 filas
print(datos_ventas.tail(10))
````

> Utilizar **head** al cargar un nuevo conjunto de datos permite verificar rápidamente que los datos se han importado correctamente antes de realizar cualquier operación.

## Diferencias entre Series y DataFrame

Aunque Series y DataFrame están relacionados, tienen características distintas que determinan cuándo usar cada uno:

```mermaid
flowchart TB
    subgraph Series
        S1[Una dimensión]
        S2[Un solo dtype]
        S3[Un índice]
        S4[Representa una columna]
    end
    subgraph DataFrame
        D1[Dos dimensiones]
        D2[Múltiples dtypes]
        D3[Índice + columnas]
        D4[Tabla completa]
    end
    Series --> |"Columna de"| DataFrame
```

| Característica | Series | DataFrame |
|---------------|--------|-----------|
| Dimensiones | 1 | 2 |
| Estructura | Vector con índice | Tabla con filas y columnas |
| Tipo de dato | Homogéneo | Heterogéneo por columna |
| Acceso | Por índice | Por índice y columna |
| Caso de uso | Variable individual | Conjunto de variables |

> Cada columna de un DataFrame es una Series. Esta relación permite extraer columnas individuales y aplicarles operaciones específicas de Series.

### Extraer Series de un DataFrame

````python
df = pd.DataFrame({
    'nombre': ['Ana', 'Carlos', 'Elena'],
    'edad': [28, 35, 42],
    'salario': [45000, 52000, 61000]
})

# Extraer una columna como Series
edades = df['edad']
print(type(edades))  # <class 'pandas.core.series.Series'>

# La Series conserva el índice del DataFrame
print(edades)
````

### Convertir Series a DataFrame

````python
# Series individual
precios = pd.Series([10.5, 15.0, 12.0], name='precio')

# Convertir a DataFrame
df_precios = precios.to_frame()
print(df_precios)
````

```
   precio
0    10.5
1    15.0
2    12.0
```

## Tipos de datos en Pandas

Pandas utiliza tipos de datos específicos que optimizan el almacenamiento y las operaciones. Conocer estos tipos permite trabajar de forma más eficiente con grandes volúmenes de datos.

### Tipos principales

| Tipo Pandas | Descripción | Ejemplo |
|-------------|-------------|---------|
| int64 | Enteros de 64 bits | 42, -15, 0 |
| float64 | Decimales de 64 bits | 3.14, -0.5 |
| object | Texto o datos mixtos | 'hola', 'mundo' |
| bool | Valores booleanos | True, False |
| datetime64 | Fechas y horas | 2025-01-15 |
| category | Categorías con valores limitados | 'bajo', 'medio', 'alto' |
| string | Texto optimizado | 'cadena' |

### Especificar tipos al crear estructuras

````python
# Series con tipo explícito
cantidades = pd.Series([1, 2, 3, 4], dtype='int32')
print(cantidades.dtype)  # int32

# DataFrame con tipos específicos
df = pd.DataFrame({
    'id': pd.Series([1, 2, 3], dtype='int32'),
    'nombre': pd.Series(['A', 'B', 'C'], dtype='string'),
    'activo': pd.Series([True, False, True], dtype='bool'),
    'nivel': pd.Series(['bajo', 'medio', 'alto'], dtype='category')  # Ejemplo de category
})
print(df.dtypes)
````

* ``string``: Para texto general donde los valores pueden ser muy variados o únicos (nombres de personas, descripciones, etc.)

* ``category``: Para datos con un conjunto limitado y repetido de valores posibles. Pandas los almacena internamente como códigos enteros, lo que ahorra memoria significativamente.

### Uso del tipo string moderno

Pandas incluye un tipo **string** dedicado que ofrece mejor rendimiento y comportamiento más consistente que el tipo object tradicional:

````python
# Crear Series con tipo string
nombres = pd.Series(['Ana', 'Carlos', 'Elena'], dtype='string')
print(nombres.dtype)  # string

# También funciona con pd.StringDtype()
apellidos = pd.Series(['García', 'López', 'Martín'], dtype=pd.StringDtype())
````

## Operaciones básicas con el índice

El índice es un componente fundamental en Pandas que permite alinear datos automáticamente durante las operaciones.

### Modificar el índice

````python
df = pd.DataFrame({
    'producto': ['Laptop', 'Monitor', 'Teclado'],
    'precio': [899, 299, 79]
})

# Establecer una columna como índice
df_indexado = df.set_index('producto')
print(df_indexado)
````

```
          precio
producto        
Laptop       899
Monitor      299
Teclado       79
```

### Restablecer el índice

````python
# Volver a índice numérico
df_reset = df_indexado.reset_index()
print(df_reset)
````

### Verificar propiedades del índice

````python
# Índice actual
print(df.index)

# Verificar si el índice es único
print(df.index.is_unique)  # True

# Verificar si está ordenado
print(df.index.is_monotonic_increasing)
````

> Mantener un índice único facilita las operaciones de búsqueda y combinación de datos. Pandas permite índices duplicados, pero esto puede generar comportamientos inesperados en ciertas operaciones.
