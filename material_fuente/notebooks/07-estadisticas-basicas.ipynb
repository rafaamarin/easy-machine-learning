# Estadísticas básicas en Pandas

El **análisis exploratorio** de datos comienza con preguntas sencillas: ¿cuántos registros tengo? ¿Cuál es el valor promedio? ¿Hay valores extremos? Pandas proporciona un conjunto de métodos estadísticos que responden estas preguntas con una sola línea de código. Estos métodos operan de forma **vectorizada**, lo que significa que calculan estadísticas sobre columnas completas sin necesidad de bucles explícitos.

## Resumen estadístico con describe

El método **describe** genera un resumen estadístico completo de las columnas numéricas del DataFrame. Es el punto de partida habitual para comprender la distribución de los datos.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'precio': [120, 89, 250, 175, 95, 310, 145, 200],
    'unidades_vendidas': [45, 120, 30, 65, 90, 15, 55, 40],
    'valoracion': [4.2, 3.8, 4.7, 4.1, 3.5, 4.9, 4.0, 4.3]
})

# Resumen estadístico de columnas numéricas
print(df.describe())
````

```
50%    160.000000          50.000000    4.150000
```

El resumen incluye:

* **min** y **max**: valores extremos
* **25%, 50%, 75%**: percentiles (cuartiles)

> El método **describe** excluye automáticamente los valores nulos del cálculo. El percentil 50% coincide con la **mediana**.

## Conteo de valores con value_counts

El método **value_counts** cuenta las ocurrencias de cada valor único en una columna. Es fundamental para analizar **variables categóricas** y detectar patrones en los datos.

````python
df = pd.DataFrame({
    'cliente': ['Ana', 'Carlos', 'Ana', 'Elena', 'Carlos', 'Ana', 'David', 'Carlos'],
    'categoria': ['Premium', 'Básico', 'Premium', 'Básico', 'Premium', 'Premium', 'Básico', 'Básico'],
    'region': ['Norte', 'Sur', 'Norte', 'Este', 'Sur', 'Norte', 'Este', 'Sur']
})

# Contar valores de una columna
print(df['categoria'].value_counts())
````

```
categoria
Premium    4
Básico     4
Name: count, dtype: int64
```

Por defecto, los resultados se ordenan de mayor a menor frecuencia.

### Opciones de value_counts

````python
# Ordenar alfabéticamente en lugar de por frecuencia
print(df['cliente'].value_counts(sort=False))

# Mostrar porcentajes en lugar de conteos
print(df['categoria'].value_counts(normalize=True))
````

```
categoria
Premium    0.5
Básico     0.5
Name: proportion, dtype: float64
```

> El parámetro **normalize=True** devuelve proporciones que suman 1.0, útiles para calcular porcentajes multiplicando por 100.

### Orden ascendente y valores nulos

````python
df_con_nulos = pd.DataFrame({
    'estado': ['Activo', 'Inactivo', None, 'Activo', 'Pendiente', 'Activo', None]
})

# Orden ascendente (menor a mayor frecuencia)
print(df_con_nulos['estado'].value_counts(ascending=True))
````

````python
# Incluir valores nulos en el conteo
print(df_con_nulos['estado'].value_counts(dropna=False))
````

## Valores únicos con unique y nunique

Los métodos **unique** y **nunique** identifican los valores distintos en una columna.

### unique: obtener valores únicos

````python
df = pd.DataFrame({
    'producto': ['Laptop', 'Monitor', 'Laptop', 'Teclado', 'Monitor', 'Mouse'],
    'marca': ['Dell', 'LG', 'HP', 'Logitech', 'Dell', 'Logitech']
})

# Obtener array de valores únicos
print(df['producto'].unique())
````

```
['Laptop' 'Monitor' 'Teclado' 'Mouse']
```

El resultado es un **array de NumPy** que contiene cada valor una sola vez, en el orden en que aparecen por primera vez.

### nunique: contar valores únicos

````python
# Número de valores únicos
print(df['producto'].nunique())  # 4
print(df['marca'].nunique())     # 4
````

> El método **nunique** equivale a `len(df['columna'].unique())`, pero es más expresivo y ligeramente más eficiente.

### nunique en todo el DataFrame

````python
# Valores únicos por columna
print(df.nunique())
````

```
producto    4
marca       4
dtype: int64
```

### Manejo de valores nulos

Por defecto, **nunique** excluye los valores nulos:

````python
df_con_nulos = pd.DataFrame({
    'categoria': ['A', 'B', None, 'A', 'C', None]
})

# Sin contar nulos (por defecto)
print(df_con_nulos['categoria'].nunique())  # 3

# Contando nulos como un valor más
print(df_con_nulos['categoria'].nunique(dropna=False))  # 4
````

## Funciones de agregación

Pandas proporciona métodos para calcular **estadísticas de resumen** sobre columnas numéricas. Estas funciones se aplican directamente sobre Series o DataFrames.

### Operaciones sobre Series

````python
df = pd.DataFrame({
    'ventas': [1500, 2300, 1800, 2100, 2500, 1900, 2200, 1700],
    'gastos': [800, 1200, 950, 1100, 1300, 900, 1150, 850],
    'clientes': [45, 68, 52, 61, 75, 55, 65, 48]
})

# Suma total
print(df['ventas'].sum())  # 16000

# Media aritmética
print(df['ventas'].mean())  # 2000.0

# Mediana (valor central)
print(df['ventas'].median())  # 2000.0

# Desviación estándar
print(df['ventas'].std())  # 333.33...

# Valor mínimo
print(df['ventas'].min())  # 1500

# Valor máximo
print(df['ventas'].max())  # 2500
````

### Operaciones sobre DataFrame

Cuando se aplican al DataFrame completo, estas funciones operan sobre **cada columna**:

````python
# Suma por columna
print(df.sum())
````

```
ventas      16000
gastos       8250
clientes      469
dtype: int64
```

````python
# Media por columna
print(df.mean())
````

```
ventas      2000.000
gastos      1031.250
clientes      58.625
dtype: float64
```

> En Pandas 2.x, las funciones de agregación usan `numeric_only=False` por defecto. Si el DataFrame contiene columnas de texto, es recomendable usar `numeric_only=True` para evitar errores.

````python
df_mixto = pd.DataFrame({
    'producto': ['A', 'B', 'C'],
    'precio': [100, 200, 150],
    'cantidad': [10, 5, 8]
})

# Especificar que solo opere sobre columnas numéricas
print(df_mixto.mean(numeric_only=True))
````

```
precio      150.000000
cantidad      7.666667
dtype: float64
```

### Tabla de funciones de agregación

| Método | Descripción | Ejemplo |
|--------|-------------|---------|
| sum() | Suma de valores | `df['col'].sum()` |
| mean() | Media aritmética | `df['col'].mean()` |
| median() | Mediana (percentil 50) | `df['col'].median()` |
| std() | Desviación estándar | `df['col'].std()` |
| var() | Varianza | `df['col'].var()` |
| min() | Valor mínimo | `df['col'].min()` |
| max() | Valor máximo | `df['col'].max()` |
| count() | Número de valores no nulos | `df['col'].count()` |

### Cálculo por filas

Por defecto, las funciones operan sobre columnas (`axis=0`). Para operar sobre filas, se usa `axis=1`:

````python
# Suma por fila (total de cada registro)
df['total'] = df[['ventas', 'gastos']].sum(axis=1)

# Media por fila
df['promedio'] = df[['ventas', 'gastos', 'clientes']].mean(axis=1)

print(df)
````

## Percentiles con quantile

El método **quantile** calcula percentiles específicos de la distribución. Un percentil indica el valor por debajo del cual se encuentra un porcentaje determinado de los datos.

````python
df = pd.DataFrame({
    'salario': [28000, 32000, 35000, 42000, 48000, 55000, 62000, 85000, 95000, 120000]
})

# Mediana (percentil 50)
print(df['salario'].quantile(0.5))  # 51500.0

# Primer cuartil (25%)
print(df['salario'].quantile(0.25))  # 36750.0

# Tercer cuartil (75%)
print(df['salario'].quantile(0.75))  # 79250.0
````

### Múltiples percentiles

````python
# Calcular varios percentiles a la vez
percentiles = df['salario'].quantile([0.1, 0.25, 0.5, 0.75, 0.9])
print(percentiles)
````

```
0.10     30800.0
0.25     36750.0
0.50     51500.0
0.75     79250.0
0.90    102500.0
Name: salario, dtype: float64
```

### Rango intercuartílico (IQR)

El **rango intercuartílico** mide la dispersión central de los datos y se utiliza para detectar valores atípicos:

````python
q1 = df['salario'].quantile(0.25)
q3 = df['salario'].quantile(0.75)
iqr = q3 - q1

# Límites para detección de outliers
limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

print(f"IQR: {iqr}")
print(f"Límites: [{limite_inferior}, {limite_superior}]")
````

```
IQR: 42500.0
Límites: [-27000.0, 143000.0]
```

> Los valores fuera del rango \([Q1 - 1.5 \times IQR, Q3 + 1.5 \times IQR]\) se consideran **valores atípicos** (outliers) según el criterio de Tukey.

Para **filtrar outliers** del DataFrame, aplicamos una máscara booleana que selecciona solo los valores dentro de los límites:

````python
# Filtrar datos sin outliers
df_sin_outliers = df[(df['salario'] >= limite_inferior) & (df['salario'] <= limite_superior)]

print(f"Registros originales: {len(df)}")
print(f"Registros sin outliers: {len(df_sin_outliers)}")
print(f"Outliers eliminados: {len(df) - len(df_sin_outliers)}")
````

````python
# Alternativa con between()
df_sin_outliers = df[df['salario'].between(limite_inferior, limite_superior)]
````

````python
# Obtener solo los outliers
outliers = df[(df['salario'] < limite_inferior) | (df['salario'] > limite_superior)]
print(outliers)
````

## Correlación entre variables

El método **corr** calcula la **correlación de Pearson** entre variables numéricas. La correlación mide la fuerza y dirección de la relación lineal entre dos variables, con valores entre -1 y 1.

````python
df = pd.DataFrame({
    'horas_estudio': [2, 4, 6, 8, 10, 3, 5, 7, 9, 1],
    'calificacion': [55, 70, 78, 88, 95, 60, 72, 82, 90, 50],
    'horas_sueno': [8, 7, 6, 5, 5, 7, 7, 6, 5, 9],
    'edad': [20, 22, 21, 23, 20, 22, 21, 23, 22, 20]
})

# Matriz de correlación
print(df.corr())
````

### Métodos de correlación

El parámetro **method** permite usar diferentes coeficientes de correlación:

````python
# Correlación de Pearson (por defecto)
df.corr(method='pearson')

# Correlación de Spearman (basada en rangos)
df.corr(method='spearman')

# Correlación de Kendall (basada en concordancia)
df.corr(method='kendall')
````

| kendall | Muestras pequeñas, datos con empates |

## Agregación personalizada con agg

El método **agg** permite aplicar **múltiples funciones** de agregación simultáneamente, ofreciendo mayor flexibilidad que los métodos individuales.

````python
df = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D', 'E'],
    'ventas': [1500, 2300, 1800, 2100, 2500],
    'margen': [15.5, 22.0, 18.5, 20.0, 25.0]
})

# Aplicar múltiples funciones a una columna
print(df['ventas'].agg(['sum', 'mean', 'std', 'min', 'max']))
````

```
sum     10200.000000
mean     2040.000000
std       377.491722
min      1500.000000
max      2500.000000
Name: ventas, dtype: float64
```

### Funciones diferentes por columna

````python
# Diferentes agregaciones para cada columna
agregaciones = df.agg({
    'ventas': ['sum', 'mean'],
    'margen': ['mean', 'max']
})
print(agregaciones)
````

```
        ventas  margen
sum   10200.0     NaN
mean   2040.0    20.2
max       NaN    25.0
```

### Funciones personalizadas

````python
# Definir función personalizada
def rango(x):
    return x.max() - x.min()

# Usar función personalizada con agg
print(df['ventas'].agg(['mean', rango]))
````

```
mean     2040.0
rango    1000.0
Name: ventas, dtype: float64
```

### Agregar nombres descriptivos

Para asignar nombres personalizados a las funciones de agregación, tenemos dos opciones principales:

````python
# Opción 1: Usando renombrado manual
resultado = df.agg({
    'ventas': ['sum', 'mean', 'std'],
    'margen': ['mean', 'max']
})

# Renombrar el índice con nombres descriptivos
resultado = resultado.rename(index={
    'sum': 'total',
    'mean': 'promedio', 
    'std': 'dispersion',
    'max': 'maximo'
})
print(resultado)
````

```
              ventas  margen
total        10200.0     NaN
promedio      2040.0    20.2
dispersion     408.2     NaN
maximo           NaN    25.0
```

````python
# Opción 2: Usando pd.NamedAgg para mayor claridad
resultado = df.agg(
    ventas_total=pd.NamedAgg(column='ventas', aggfunc='sum'),
    ventas_promedio=pd.NamedAgg(column='ventas', aggfunc='mean'),
    ventas_dispersion=pd.NamedAgg(column='ventas', aggfunc='std'),
    margen_medio=pd.NamedAgg(column='margen', aggfunc='mean'),
    margen_maximo=pd.NamedAgg(column='margen', aggfunc='max')
)
print(resultado)
````

```
ventas_total         10200.0
ventas_promedio       2040.0
ventas_dispersion      408.2
margen_medio            20.2
margen_maximo           25.0
dtype: float64
```

> El método **agg** acepta nombres de funciones como strings (`'sum'`, `'mean'`), funciones de NumPy (`np.sum`), o funciones definidas por el usuario. Para asignar nombres personalizados, usa `pd.NamedAgg` o renombra el índice después de la agregación.

## Estadísticas adicionales

Pandas incluye métodos estadísticos adicionales para análisis más específicos.

### Moda

La **moda** es el valor que más se repite:

````python
df = pd.DataFrame({
    'talla': ['M', 'L', 'M', 'S', 'M', 'L', 'XL', 'M']
})

# Moda (puede haber múltiples)
print(df['talla'].mode())
````

### Asimetría y curtosis

````python
df = pd.DataFrame({
    'valores': [10, 12, 14, 15, 16, 18, 20, 50, 80]
})

# Asimetría (skewness): mide la simetría de la distribución
print(f"Asimetría: {df['valores'].skew():.4f}")

# Curtosis: mide la concentración de datos en las colas
print(f"Curtosis: {df['valores'].kurtosis():.4f}")
````

```
Asimetría: 1.5892
Curtosis: 1.6129
```

| Asimetría | Interpretación |
|-----------|----------------|
| = 0 | Distribución simétrica |
| > 0 | Cola derecha más larga (sesgo positivo) |
| < 0 | Cola izquierda más larga (sesgo negativo) |

### Acumulados

Las funciones acumulativas calculan estadísticas progresivas:

````python
df = pd.DataFrame({
    'mes': ['Ene', 'Feb', 'Mar', 'Abr', 'May'],
    'ventas': [100, 150, 120, 180, 200]
})

# Suma acumulada
df['ventas_acumuladas'] = df['ventas'].cumsum()

# Máximo acumulado
df['maximo_historico'] = df['ventas'].cummax()

# Media acumulada
df['media_movil'] = df['ventas'].expanding().mean()

print(df)
````

```
   mes  ventas  ventas_acumuladas  maximo_historico  media_movil
0  Ene     100                100               100       100.00
```

## Ejemplo práctico de análisis exploratorio

Un flujo típico de análisis exploratorio combina varios métodos estadísticos:

````python
import pandas as pd
import numpy as np

# Datos de ejemplo: ventas de una tienda
np.random.seed(42)
df = pd.DataFrame({
    'fecha': pd.date_range('2025-01-01', periods=100),
    'categoria': np.random.choice(['Electrónica', 'Ropa', 'Hogar', 'Deportes'], 100),
    'ventas': np.random.randint(50, 500, 100),
    'unidades': np.random.randint(1, 20, 100),
    'descuento': np.random.choice([0, 5, 10, 15, 20], 100)
})

# 1. Visión general
print("=== Resumen estadístico ===")
print(df.describe())

# 2. Distribución de categorías
print("\n=== Ventas por categoría ===")
print(df['categoria'].value_counts())

# 3. Estadísticas por grupo
print("\n=== Ventas promedio por categoría ===")
print(df.groupby('categoria')['ventas'].agg(['mean', 'median', 'std']))

# 4. Correlaciones
print("\n=== Correlación ventas-unidades ===")
print(f"Correlación: {df['ventas'].corr(df['unidades']):.4f}")

# 5. Percentiles de ventas
print("\n=== Distribución de ventas ===")
print(df['ventas'].quantile([0.25, 0.5, 0.75, 0.9]))
````

> Este flujo de análisis responde preguntas clave: ¿cómo se distribuyen los datos? ¿hay diferencias entre grupos? ¿existen relaciones entre variables? ¿dónde están los valores extremos?
